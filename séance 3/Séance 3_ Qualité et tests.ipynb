# **1. Définition et concept**

**<h1>1. Qualité et tests dans la maintenance applicative</h1>**

Maintenir une application ne consiste pas seulement à corriger des bugs. Il faut aussi **garantir que le logiciel reste fiable dans le temps**, même après des modifications.

Chaque correction, amélioration ou adaptation technique peut introduire de nouveaux problèmes. C’est pourquoi la **qualité logicielle** est un pilier de la maintenance.

**<h2>1.1 Pourquoi la qualité est essentielle en maintenance ?</h2>**

Lorsqu’un logiciel évolue :

* On modifie du code existant
* On ajoute des fonctionnalités
* On corrige des erreurs
* On adapte l’application à de nouveaux environnements

Sans démarche qualité, on risque de **casser ce qui fonctionnait déjà**.

**<h2>1.2 Objectifs</h2>**

À la fin, les étudiants doivent comprendre :

* Comment **repérer et analyser une erreur**
* Comment **suivre le comportement d’une application**
* Comment **vérifier automatiquement que le code fonctionne**
* Pourquoi les tests sont indispensables après chaque modification
* Le lien direct entre **tests et maintenance durable**

---


**<h1>2. Débogage</h1>**

**<h2>2.1 Définition</h2>**

Le **débogage** est une méthode permettant de comprendre pourquoi un programme ne fonctionne pas comme prévu et d’identifier l’origine d’une erreur.

C’est souvent la **première étape en maintenance corrective**.

On ne corrige pas un bug au hasard :
on observe → on analyse → on corrige.


**<h2>2.2 Exemple pédagogique</h2>**

Objectif : empêcher l’ajout d’une tâche vide dans une TodoList.


````python
def add_task(tasks, task):
    print("Tâche reçue :", task)  # trace pour comprendre ce qui arrive dans la fonction

    if task.strip() == "":
        print("Erreur : tâche vide")
        return False

    tasks.append(task)
    print("Tâche ajoutée")
    return True


tasks = []

add_task(tasks, "")
print("Liste finale :", tasks)
````

**<h2>2.3 Pourquoi cet exemple est clair ?</h2>**

On affiche :

* la valeur reçue
* la décision prise
* l’état final

Cela permet de **suivre l’exécution pas à pas**.

**<h2>2.4 Lien avec la maintenance</h2>**

En situation réelle :

* un utilisateur signale un bug
* le mainteneur reproduit le problème
* il observe le programme
* il identifie la cause

Le débogage est donc une **compétence centrale du mainteneur**.

---


**<h1>3. Logs (journalisation)</h1>**

**<h2>3.1 Définition</h2>**

Les **logs** sont des messages enregistrés automatiquement par l’application pour suivre son activité.

Contrairement aux `print`, ils sont faits pour les environnements professionnels.

On peut enregistrer :

* erreurs
* connexions
* actions utilisateurs
* démarrage du système


**<h2>3.2 Exemple clair</h2>**


````python
# Mettre le code dans un ficher .py pour faire le test

import logging

logging.basicConfig(level=logging.INFO)

def add_task(tasks, task):
    if task.strip() == "":
        logging.warning("Tentative d'ajout d'une tâche vide")
        return False

    tasks.append(task)
    logging.info("Tâche ajoutée : %s", task)
    return True


tasks = []

add_task(tasks, "Réviser le cours")
add_task(tasks, "")
````

**<h2>3.3 Que va-t-il se passer ?</h2>**

Console :

* INFO → tâche ajoutée
* WARNING → tentative incorrecte

Même sans voir l’utilisateur, on comprend ce qu’il s’est passé.

**<h2>3.4 Lien avec la maintenance</h2>**

Très souvent, le mainteneur **n’a pas accès directement à l’utilisateur**.

Les logs deviennent alors indispensables pour :

* diagnostiquer un incident
* comprendre un crash
* analyser un comportement anormal

Sans logs = maintenance beaucoup plus difficile.

**<h1>4. Tests unitaires</h1>**

**<h2>4.1 Définition</h2>**

Un **test unitaire** vérifie qu’une fonction fonctionne correctement **prise isolément**.

On compare :

résultat attendu vs résultat obtenu.

**<h2>4.2 Fonction à tester</h2>**

Règles métier :

* Une tâche vide → refusée
* Une tâche valide → ajoutée


````python
def add_task(tasks, task):
    if task.strip() == "":
        return False

    tasks.append(task)
    return True
````

**<h2>4.3 Tests</h2>**


````python
def test_add_valid_task():
    tasks = []
    result = add_task(tasks, "Réviser Python")

    assert result == True
    assert tasks == ["Réviser Python"]


def test_add_empty_task():
    tasks = []
    result = add_task(tasks, "")

    assert result == False
    assert tasks == []


test_add_valid_task()
test_add_empty_task()

print("Tests unitaires OK")
````

**<h2>4.4 Pourquoi c’est important en maintenance ?</h2>**

Imagine :

* tu modifies la fonction 6 mois plus tard
* tu lances les tests
* un test échoue

Tu sais immédiatement que ta modification a cassé un comportement.

Les tests sont une **ceinture de sécurité du mainteneur**.

---


**<h1>5. Tests d’intégration</h1>**

**<h2>5.1 Définition</h2>**

Les tests d’intégration vérifient que **plusieurs fonctions fonctionnent correctement ensemble**.

Car une application n’est jamais composée d’une seule fonction.

**<h2>5.2 Exemple</h2>**


````python
def remove_task(tasks, task):
    if task in tasks:
        tasks.remove(task)
        return True
    return False

def test_task_lifecycle():
    tasks = []

    add_task(tasks, "Faire le TP")
    remove_task(tasks, "Faire le TP")

    assert tasks == []


test_task_lifecycle()
print("Test d'intégration OK")
````

On teste ici un **scénario réel** :

Ajouter → supprimer → vérifier l’état final.

**<h2>5.3 Lien avec la maintenance</h2>**

Après une mise à jour :

* une fonction peut marcher seule
* mais casser une interaction

Les tests d’intégration évitent ce type de surprise.

---


**<h1>6. Tests de régression</h1>**

**<h2>6.1 Définition</h2>**

Un test de régression vérifie qu’une modification **n’a pas détérioré une fonctionnalité existante**.

Très critique en maintenance évolutive.


**<h2>6.2 Cas réaliste</h2>**

On décide d’améliorer l’application pour **refuser les doublons**.

Nouvelle règle :

* Impossible d’ajouter deux fois la même tâche.



````python
def add_task(tasks, task):
    if task.strip() == "":
        return False

    if task in tasks:
        return False

    tasks.append(task)
    return True
````

**<h2>6.3 Test de régression</h2>**

````python
def test_no_duplicate():
    tasks = ["Réviser Python"]

    result = add_task(tasks, "Réviser Python")

    assert result == False
    assert tasks == ["Réviser Python"]


test_no_duplicate()
print("Test de régression OK")
````

Si ce test échoue après une modification future, on saura que la règle métier a été cassée.

---


**<h1>7. À retenir</h1>**

Dans la maintenance applicative :

* Le **débogage** sert à trouver les causes d’un problème
* Les **logs** permettent de comprendre ce qui se passe en production
* Les **tests unitaires** sécurisent les fonctions
* Les **tests d’intégration** sécurisent les interactions
* Les **tests de régression** protègent l’existant

Ensemble, ils garantissent une **maintenance fiable, maîtrisée et professionnelle**.

# **3. Bonus - Automatiser le test sur Github**

**<h1>Approche automatisée des tests sur la TodoList Streamlit</h1>**

Cette section montre comment structurer une application pour qu’elle soit testable et comment automatiser l’exécution des tests avec GitHub Actions.

Principes :

* Séparer l’interface et la logique métier
* Tester uniquement la logique
* Automatiser les tests pour sécuriser les évolutions

Ce n’est pas un TP mais un exemple pour aller plus loin avec une pratique professionnelle utilisée en maintenance applicative.

---

**<h2>1. Extraire la logique métier</h2>**

Une interface Streamlit ne doit pas contenir les règles métier importantes.
On place les fonctions dans un fichier dédié afin de pouvoir les tester indépendamment.

**<h3>Créer les fichiers</h3>**

Dans le terminal :

```bash
touch app_functions.py
touch test_app.py
```

---

**<h3>app_functions.py</h3>**

```python
def add_task(tasks, task):
    if task.strip() == "":
        return False
    if any(t["task"] == task for t in tasks):
        return False
    tasks.append({"task": task, "done": False})
    return True


def mark_done(tasks, index):
    if 0 <= index < len(tasks):
        tasks[index]["done"] = True
        return True
    return False
```

Ces fonctions deviennent le **cœur métier** de l’application.
Elles peuvent maintenant être testées sans lancer Streamlit.

---

**<h2>2. Utiliser ces fonctions dans app.py</h2>**

L’application ne doit plus contenir la logique d’ajout ou de modification.

**<h3>Importer les fonctions</h3>**

Au début de **app.py** :

```python

```

Avant (logique dans Streamlit) → à éviter.

Après :

Si un test échoue après une modification, on sait immédiatement que le comportement métier est cassé.

**<h2>4. Automatiser les tests avec GitHub Actions</h2>**

L’objectif est simple :
**chaque push ou Pull Request déclenche automatiquement les tests.**

Si un test échoue, la modification ne doit pas être fusionnée.

---

**<h3>Créer le workflow</h3>**

1. Ouvrir votre dépôt GitHub
2. Aller dans l’onglet **Actions**
3. Cliquer sur **set up a workflow yourself**
4. GitHub crée automatiquement le dossier :

```
.github/workflows/
```

5. Créer le fichier :

```
python-tests.yml
```

---

**<h3>Contenu du fichier YAML</h3>**

```yaml
name: Python Tests

on:
  push:
    branches: [ main, feature/tests ]
  pull_request:
    branches: [ main ]

jobs:
  test:
    runs-on: ubuntu-latest

    steps:
    - uses: actions/checkout@v3

    - name: Setup Python
      uses: actions/setup-python@v4
      with:
        python-version: '3.10'

    - name: Install dependencies
      run: |
        pip install pytest
        pip install streamlit

    - name: Run tests
      run: |
        pytest test_app.py
```

---

**<h2>5. Workflow Git recommandé</h2>**

Créer une branche dédiée :

```bash
git checkout -b feature/tests
```

Ajouter les fichiers :

```bash
git add app_functions.py test_app.py .github/workflows/python-tests.yml
git commit -m "Ajout des tests et du workflow GitHub Actions"
```

Envoyer la branche :

```bash
git push origin feature/tests
```

Créer ensuite une **Pull Request** vers `main`.

---

**<h2>6. Vérifier l’exécution automatique</h2>**

Sur GitHub :

* Onglet **Actions**
* Job **Python Tests**

Deux cas :

* Tests réussis → la Pull Request peut être fusionnée
* Tests en échec → correction obligatoire avant fusion

Cette approche permet :
