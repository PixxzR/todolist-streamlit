# **1. Comprendre un code non écrit par soi**

En maintenance applicative, la situation la plus fréquente n’est pas d’écrire une application, mais de reprendre celle de quelqu’un d’autre.

Avant toute modification, une règle fondamentale s’impose :

**On ne modifie jamais un code que l’on ne comprend pas.**

Changer un code sans en maîtriser la logique est l’une des principales causes de régression en production.

---

**<h2>Objectif</h2>**

Apprendre à analyser rapidement un code existant pour :

* comprendre son fonctionnement
* identifier ses responsabilités
* repérer les zones à risque
* préparer un futur refactoring

Il ne s’agit pas encore d’améliorer le code — uniquement de **le lire comme un mainteneur professionnel**.

---


**<h2>La bonne posture du mainteneur</h2>**

Lorsqu’on découvre un projet, il faut éviter deux erreurs classiques :

**Erreur n°1 : plonger directement dans les modifications**
Très risqué.

**Erreur n°2 : juger le code trop vite**
Il peut répondre à des contraintes que vous ignorez.

Un mainteneur agit comme un enquêteur : il observe avant d’agir.

---



**<h2>Méthode simple pour lire un code inconnu</h2>**

Toujours analyser dans cet ordre.

**Que permet ce fichier ?**

Exemples :

* manipuler des données
* appeler une API

Sans cette vision globale, on se perd dans les détails.


**<h3>2. Observer les noms</h3>**

Les noms sont la première documentation d’un code.

Un bon nom doit permettre de deviner le rôle sans lire l’implémentation.

À examiner :

* variables

* structures de données

Si vous ne comprenez pas un nom en moins de deux secondes, il existe probablement un problème de lisibilité.



**<h3>3. Identifier les responsabilités</h3>**

Un fichier propre doit répondre à une idée simple :

**Une responsabilité principale.**

Exemples :

* afficher une interface
* gérer la logique métier

Lorsque tout est mélangé, la maintenance devient difficile.


**<h3>4. Repérer les signaux d’alerte</h3>**

Certains indices doivent immédiatement attirer votre attention :

* variables aux noms trop courts
* fonctions très longues
* logique dupliquée
* structures de données peu lisibles
* absence de fonctions
* dépendance forte à un framework

Ce ne sont pas encore des corrections — seulement des observations.

# **2. Refactoring : améliorer sans changer la fonctionnalité**


Après avoir appris à lire et analyser un code existant, l’étape suivante en maintenance applicative consiste à **améliorer ce code sans modifier ce qu’il fait pour l’utilisateur**.

C’est exactement le rôle du refactoring.

Un mainteneur ne réécrit pas une application parce qu’elle est “mal faite”.
Il la fait évoluer progressivement pour la rendre :

* plus lisible
* plus testable
* plus modulaire
* moins risquée à modifier

---

**<h2>Définition du refactoring</h2>**

Le refactoring consiste à **modifier la structure interne d’un code sans changer son comportement observable**.

Autrement dit :

* L’utilisateur ne voit aucune différence.
* L’application fonctionne comme avant.
* Mais le code devient plus simple à maintenir.

C’est une pratique centrale en maintenance moderne.

---


**<h2>Pourquoi refactoriser ?</h2>**

Avec le temps, un projet accumule ce que l’on appelle de la **dette technique** :

Un code qui fonctionne, mais qui devient :

* difficile à comprendre
* long à modifier
* propice aux bugs
* coûteux à faire évoluer

Refactoriser permet de réduire cette dette avant qu’elle ne bloque le projet.

Important :
On ne refactorise pas “pour faire joli”.
On refactorise pour **sécuriser les futures modifications**.

---

**<h2>Règle fondamentale du refactoring</h2>**

Toujours appliquer ce principe :

**Comprendre → sécuriser → améliorer**

Jamais l’inverse.

Dans un contexte professionnel, on évite de refactoriser un code sans :

* l’avoir analysé
* avoir identifié ses faiblesses
* comprendre les impacts possibles

---


**<h2>Retour sur le code analysé précédemment</h2>**

Dans la section précédente, plusieurs problèmes ont été identifiés :

* noms de variables peu explicites
* structure de données difficile à lire
* absence de fonctions
* mélange entre interface et logique métier
* code impossible à tester

L’objectif n’est donc pas de changer ce que fait l’application, mais de **corriger ces faiblesses structurelles**.

---



**<h2>Étape 1 — Clarifier le code existant</h2>**

Avant même de séparer les fichiers, une première amélioration consiste à rendre le code compréhensible.

Exemple : remplacer des noms ambigus.

Au lieu de :

```python
if "t" not in st.session_state:
    st.session_state["t"] = []
```

On écrit :

```python
if "tasks" not in st.session_state:
    st.session_state["tasks"] = []
```

Même amélioration pour la structure :

Au lieu de :

```python
[a, False]
```

On utilise un dictionnaire :

```python
{
    "task": task_name,
    "done": False
}
```

Pourquoi ?

Parce qu’un dictionnaire est **auto-documenté**.
On comprend immédiatement ce que représente chaque valeur.

Résultat : moins d’effort mental pour le mainteneur.

---



**<h2>Étape 2 — Isoler la logique métier</h2>**

Un principe majeur en maintenance :

**L’interface doit appeler la logique — pas la contenir.**

Si toute la logique est dans Streamlit :

* impossible de tester sans lancer l’interface
* corrections plus risquées
* duplication probable

La solution consiste à déplacer la logique dans un module dédié.


**<h3>Création d’un fichier dédié à la logique</h3>**

Fichier : `app_functions.py`

```python
def init_tasks(session_state):
    if "tasks" not in session_state:
        session_state["tasks"] = []


def add_task(session_state, task):
    if task.strip() == "":
        return False
    
    session_state["tasks"].append({
        "task": task,
        "done": False
    })
    return True


def mark_done(session_state, index):
    if 0 <= index < len(session_state["tasks"]):
        session_state["tasks"][index]["done"] = True
        return True
    
    return False
```

**Ce que ce refactoring apporte :**

* chaque fonction a une responsabilité claire
* la logique devient testable
* le code est réutilisable
* les futures corrections sont localisées

On passe d’un bloc monolithique à une architecture plus professionnelle.

---



**<h2>Étape 3 — Simplifier le fichier principal</h2>**

Le rôle du fichier principal n’est plus de gérer la logique.

Il doit seulement :

* afficher l’interface
* récupérer les actions utilisateur
* appeler les fonctions

Fichier : `main.py`

```python
import streamlit as st
from app_functions import init_tasks, add_task, mark_done

init_tasks(st.session_state)

st.title("Ma TodoList (refactorisée)")

new_task = st.text_input("Ajouter une tâche")

if st.button("Ajouter"):
    add_task(st.session_state, new_task)

st.subheader("Liste des tâches")

for i, t in enumerate(st.session_state["tasks"]):
    col1, col2 = st.columns([0.8, 0.2])

    with col1:
        prefix = "✅ " if t["done"] else "🔲 "
        st.write(prefix + t["task"])

    with col2:
        if st.button("Marquer comme fait", key=f"done_{i}"):
            mark_done(st.session_state, i)
```

---



**<h2>Ce que l’utilisateur voit</h2>**

Rien ne change :

* il peut ajouter une tâche
* marquer une tâche comme faite
* voir la liste

Mais côté mainteneur, le gain est majeur.

---



**<h2>Bénéfices concrets pour la maintenance</h2>**

Après refactoring :

* le code est plus lisible
* les responsabilités sont séparées
* les tests deviennent possibles
* les bugs sont plus faciles à corriger
* les évolutions sont moins risquées

Le projet devient maintenable dans la durée.

C’est l’objectif principal d’un mainteneur.

---



**<h2>Attention à une idée reçue</h2>**

Refactoriser ne signifie pas tout réécrire.

En entreprise, on privilégie :

**le refactoring progressif.**

Pourquoi ?

Parce que réécrire entièrement une application :

* introduit souvent de nouveaux bugs

* crée de l’instabilité

On améliore donc le code **petit à petit**, à chaque intervention.

**<h2>Ce qu’il faut retenir</h2>**

Le refactoring est une compétence essentielle en maintenance applicative.

Il repose sur quelques principes simples :

* ne jamais modifier un code non compris

Un bon mainteneur ne cherche pas le code parfait.
Il cherche un code **suffisamment clair pour être modifié sans danger**.


# **3. Outils d'analyse**

En maintenance applicative, il est essentiel de **vérifier la qualité du code avant toute modification**.
Les outils d’analyse permettent de détecter automatiquement :

* erreurs potentielles
* non-conformité aux conventions de style
* complexité inutile
* duplications

Ces analyses facilitent le refactoring et sécurisent les modifications.

---


**<h2>Linter : flake8</h2>**

`flake8` est un outil qui **analyse le code Python pour détecter les erreurs et incohérences**.

Installation et utilisation :

```bash
pip install flake8
flake8 app.py
```

Exemple de sortie :

```
app.py:5:1: F401 'os' imported but unused
app.py:10:80: E501 line too long (82 > 79 characters)
```

**Interprétation :**

* `F401 'os' imported but unused` → le module `os` est importé mais jamais utilisé. Cela alourdit le code et peut induire en erreur.
* `E501 line too long (82 > 79 characters)` → la ligne dépasse la limite recommandée (PEP8). Les lignes trop longues sont difficiles à lire et à maintenir.

**Action à faire :**

* Supprimer les imports inutilisés.
* Casser les lignes trop longues en plusieurs lignes logiques.

Le linter ne corrige pas automatiquement, mais **indique précisément où intervenir**.

---

**<h2>Formatteur automatique : black</h2>**

`black` permet de **réécrire le code Python pour respecter un style uniforme**.
Il harmonise l’indentation, la longueur des lignes et l’espacement.

Installation et utilisation :

```bash
pip install black
black app.py
```

**Ce que black fait concrètement :**

* Aligne les arguments d’une fonction sur plusieurs lignes si nécessaire
* Ajoute des espaces autour des opérateurs
* Uniformise les sauts de ligne

Exemple avant / après :

Avant :

```python
if task.strip()=="":
    return False
```

Après black :

```python
if task.strip() == "":
    return False
```

**Bénéfice :** un code homogène est **plus facile à lire, comprendre et maintenir**, surtout lorsqu’il est modifié par plusieurs personnes.

---



**<h2>Analyse de qualité : SonarQube / SonarCloud (optionnel)</h2>**

Pour aller plus loin, certaines entreprises utilisent des outils comme **SonarQube** ou **SonarCloud**.

Ils détectent automatiquement :

* **Bugs potentiels** → erreurs logiques ou risques d’exceptions
* **Vulnérabilités** → failles de sécurité dans le code
* **Code smells** → code complexe, dupliqué ou difficile à maintenir

**Intérêt pour la maintenance :**

* Identifier les zones critiques avant de les modifier
* Mesurer la qualité du code dans le temps
* Suivre les corrections et le refactoring de manière systématique

Même si SonarQube est optionnel pour un TP, il est **couramment utilisé dans les projets professionnels** pour sécuriser les évolutions.

---



**<h2>Résumé</h2>**

Avant de modifier un code existant, un mainteneur doit :

1. Lancer un **linter** pour corriger les erreurs et incohérences de style.
2. Utiliser un **formatteur automatique** pour homogénéiser le code.
3. (Optionnel) Analyser la qualité globale avec un outil comme **SonarQube**.

Ces étapes permettent de réduire le risque d’introduire des bugs lors du refactoring et de rendre le code plus lisible pour toute l’équipe.


# **4. TP : Étude & Refactoring de TodoList**

L’objectif du TP est de **mettre en pratique le refactoring** sur un code existant, sans changer la fonctionnalité observable, et de vérifier qu’aucune régression n’est introduite.

**<h2>Étape 1 – Démarrer avec la version initiale du code</h2>**

Chaque étudiant commence avec le fichier `todolist_bad.py` :

```python
import streamlit as st

if "t" not in st.session_state:
    st.session_state["t"] = []

st.title("Ma liste")
a = st.text_input("Ajouter une tâche")
if st.button("OK"):
    st.session_state["t"].append([a, False])

for i, x in enumerate(st.session_state["t"]):
    c1, c2 = st.columns([0.8, 0.2])
    with c1:
        st.write(x[0])
    with c2:
        if st.button("✓", key=f"d_{i}"):
            st.session_state["t"][i][1] = True
```

**Analyse à faire :**

2. Repérer la logique métier mélangée à l’affichage Streamlit.
3. Noter les structures de données peu claires (`[a, False]`).
4. Détecter les risques pour la maintenance (tests impossibles, duplication, lisibilité faible).

**Questions à se poser :**

* Que fait exactement ce fichier ?
* Quels éléments sont critiques pour le fonctionnement correct ?

**<h2>Étape 2 – Identifier la dette technique</h2>**

Sur la base de l’analyse :

* Mauvais noms de variables → difficile de comprendre le rôle de chaque élément.
* Structure de données peu lisible → `[a, False]` nécessite de mémoriser les indices.
* Mélange logique / interface → tests impossibles sans lancer Streamlit.
* Absence de fonctions → impossibilité de réutiliser ou de tester isolément.

Cette étape correspond à **repérer les zones à risque et préparer le refactoring**, comme expliqué dans la section 1 “Comprendre un code non écrit par soi”.

---

**<h2>Étape 3 – Refactoring progressif</h2>**

**Objectif :** séparer la logique métier de l’interface et améliorer la lisibilité.

**3.1 Créer `app_functions.py`**

```python
# app_functions.py
def init_tasks(session_state):
    if "tasks" not in session_state:
        session_state["tasks"] = []

def add_task(session_state, task):
    if task.strip() and not any(t["task"] == task for t in session_state["tasks"]):
        session_state["tasks"].append({"task": task, "done": False})
        return True
    return False

def mark_done(session_state, index):
    if 0 <= index < len(session_state["tasks"]):
        session_state["tasks"][index]["done"] = True
        return True
    return False
```

**3.2 Modifier `main.py` pour utiliser les fonctions**

```python
import streamlit as st
from app_functions import init_tasks, add_task, mark_done

init_tasks(st.session_state)

st.title("Ma TodoList refactorisée")

new_task = st.text_input("Ajouter une tâche")
if st.button("Ajouter"):
    add_task(st.session_state, new_task)

st.subheader("Liste des tâches")
for i, t in enumerate(st.session_state["tasks"]):
    col1, col2 = st.columns([0.8, 0.2])
    with col1:
        st.write(("✅ " if t["done"] else "🔲 ") + t["task"])
    with col2:
        if st.button("Marquer comme fait", key=f"done_{i}"):
            mark_done(st.session_state, i)
```

**Points importants :**

* Les fonctions isolent la logique → testables sans Streamlit.
* Les noms sont explicites (`tasks`, `add_task`, `mark_done`).
* La structure des données est claire et compréhensible.

2. Lancer `black` pour uniformiser le style :

**Bénéfice :** le code est maintenant **lisible, homogène et maintenable**.

**<h2>Étape 5 – Tests unitaires pour sécuriser la logique</h2>**

Créer `test_app.py` :

```python
from app_functions import add_task, mark_done

def test_add_task():
    state = {"tasks": []}
    assert add_task(state, "Faire TD") == True
    assert state["tasks"][0]["task"] == "Faire TD"

def test_add_empty_task():
    state = {"tasks": []}
    assert add_task(state, "") == False

def test_mark_done():
    state = {"tasks": [{"task": "Faire TD", "done": False}]}
    assert mark_done(state, 0) == True
    assert state["tasks"][0]["done"] == True
```

**<h2>Étape 6 – Intégration avec GitHub Actions (optionnel)</h2>**

Automatiser l’exécution des tests et du linting lors de chaque push ou Pull Request (Se référer à la section 3 de la séance 3).

Exemple de workflow `.github/workflows/python-tests.yml` :

```yaml
name: Python Tests

on:
  push:
    branches: [ main, feature/refactor ]
  pull_request:
    branches: [ main ]

jobs:
  test:
    runs-on: ubuntu-latest

    steps:
    - uses: actions/checkout@v3

    - name: Setup Python
      uses: actions/setup-python@v4
      with:
        python-version: '3.10'

    - name: Install dependencies
      run: |
        pip install pytest flake8 black streamlit

    - name: Lint with flake8
      run: flake8 .

    - name: Format with black
      run: black --check .

    - name: Run tests
      run: pytest test_app.py
```

**Avantages :**

* Chaque modification est vérifiée automatiquement.
* Les erreurs de style ou de logique sont détectées avant fusion.
* Sécurité renforcée pour la maintenance collaborative.
